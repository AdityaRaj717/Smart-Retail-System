```python
import splitfolders
import os

# --- Configuration ---

# Set the path to your main dataset folder
# This is the folder that contains your product subfolders (e.g., 'apple', 'banana')
input_folder = 'capstone/' 

# Set the path where you want to save the split data
# A new folder 'processed_data' will be created here
output_folder = 'processed_data/' 

# Define the split ratio (90% for training, 10% for testing)
# The tuple must sum to 1.0
split_ratio = (0.9, 0.1) 

# --- End of Configuration ---


# Check if the input folder exists
if not os.path.exists(input_folder):
    print(f"Error: Input folder '{input_folder}' not found.")
else:
    print(f"Reading data from: {input_folder}")
    print(f"Output will be saved to: {output_folder}")
    print(f"Splitting data with a {int(split_ratio[0]*100)}/{int(split_ratio[1]*100)} ratio.")

    # The main function from split-folders
    # It will create 'train' and 'test' subfolders in the output directory
    # and distribute the images from each product class accordingly.
    splitfolders.ratio(
        input_folder, 
        output=output_folder, 
        seed=42, # A seed for reproducibility
        ratio=split_ratio, 
        group_prefix=None # No file grouping
    )
    
    print("\n✅ Splitting complete!")
    print(f"Your training and testing data are now available in the '{output_folder}' directory.")
```
